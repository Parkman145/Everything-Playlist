```python
import PIL.Image
import os
from math import sqrt, ceil, floor
```

```python
goal_size = 1000

image_base_path = "images"
image_names = os.listdir(image_base_path)
n_rows = ceil(sqrt(len(image_names)))
sub_image_res = floor(goal_size/n_rows)
size = n_rows*sub_image_res
row = column = 0

cover = white = PIL.Image.new("RGB", (size, size), "white")

for image_name in image_names:
    image_path = os.path.join(image_base_path, image_name)
    with PIL.Image.open(image_path) as image:
        resized = image.resize((sub_image_res, sub_image_res))
        cover.paste(resized, (row*sub_image_res, column*sub_image_res))
    row +=1
    if row >= n_rows:
        row = 0
        column += 1

PIL.Image._show(cover)
```
